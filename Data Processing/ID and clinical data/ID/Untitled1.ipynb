```python
# 2. 使用 pandas 的 merge 函数进行合并
merged_data = pd.merge(gse_data,gpl_df , left_on='ID_REF', right_on='ID', how='left')
```

```python
merged_data
```

```python
print(merged_data.columns.tolist())
```

```python
# 删除指定的列
columns_to_drop =['ID_REF','ID', 'GB_ACC', 'SPOT_ID', 'Species Scientific Name', 'Annotation Date', 'Sequence Type', 'Sequence Source', 'Target Description', 'Representative Public ID', 'Gene Title',  'ENTREZ_GENE_ID', 'RefSeq Transcript ID', 'Gene Ontology Biological Process', 'Gene Ontology Cellular Component', 'Gene Ontology Molecular Function']
df = merged_data.drop(columns=columns_to_drop)
```

```python
df
```

```python
# 4. 将 symbol 列移到最前面
cols = ['Gene Symbol'] + [col for col in df if col != 'Gene Symbol']
merged_data = df[cols]
```

```python
merged_data
```

```python
# 删除 symbol 列为 NaN 的行
merged_data = merged_data.dropna(subset=['Gene Symbol'])
```

```python
merged_data
```

```python
# 计算每个探针的表达值平均值
merged_data['mean_expression'] = merged_data.filter(regex='^GSM').mean(axis=1)

# 根据 symbol 和 mean_expression 对数据进行排序, 同一个 symbol 中 mean_expression 最大的排在前面
merged_data = merged_data.sort_values(['Gene Symbol', 'mean_expression'], ascending=[True, False])

# 根据 symbol 去重，保留 mean_expression 最大的
merged_data = merged_data.drop_duplicates(subset='Gene Symbol', keep='first')

# 删除 mean_expression 列
merged_data = merged_data.drop('mean_expression', axis=1)
```

```python
merged_data
```

```python
df = merged_data.set_index(["Gene Symbol"])
```
