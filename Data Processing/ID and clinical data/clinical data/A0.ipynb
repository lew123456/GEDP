```python
import pandas as pd
```

```python
# 设置输入和输出文件路径
input_file = 'GSE72267_series_matrix.txt'
output_file = 'GSE72267_sample_info.csv'
```

```python
def extract_sample_metadata(file_path):
    """
    从 GEO 的 series_matrix.txt 文件中提取所有 !Sample 开头的元数据行，
    并解析为字典：{字段名: [样本1值, 样本2值, ...]}
    """
    sample_metadata = {}
    
    with open(file_path, 'r', encoding='utf-8') as f:
        for line in f:
            if line.startswith('!Sample'):
                # 按制表符分割：第一列为字段名，其余为各样本的值
                parts = line.strip().split('\t')
                field_name = parts[0]
                raw_values = parts[1:]
                
                # 清理每个值：去除首尾双引号和空格
                cleaned_values = []
                for val in raw_values:
                    val = val.strip()
                    if val.startswith('"') and val.endswith('"'):
                        val = val[1:-1]  # 去掉外层双引号
                    cleaned_values.append(val)
                
                sample_metadata[field_name] = cleaned_values
                
    return sample_metadata

# 执行提取
metadata_dict = extract_sample_metadata(input_file)

# 打印摘要信息
print("✅ 成功提取以下样本字段：")
for field, values in metadata_dict.items():
    print(f"  - {field}: {len(values)} 个样本")
```

```python
# 将字典转为 DataFrame（每行 = 一个样本，每列 = 一个字段）
df_samples = pd.DataFrame(metadata_dict)

# （可选）美化列名：去掉 "!Sample " 前缀
df_samples.columns = df_samples.columns.str.replace(r'^!Sample\s*', '', regex=True)

# 保存为 CSV 文件（不保存行索引）
df_samples.to_csv(output_file, index=False)

print(f"\n✅ 样本信息已保存至: {output_file}")
print(f"📊 数据形状: {df_samples.shape[0]} 个样本, {df_samples.shape[1]} 个字段")
```
